```python
import matplotlib.pyplot as plt
```

```python
x = df['order_created_day'].unique()
y = df[df['order_status'] == 'CANCELED'].groupby(
    'order_created_day').count().reset_index()['order_status'].to_numpy()
y2 = df[df['order_status'] == 'FINISHED'].groupby(
    'order_created_day').count().reset_index()['order_status'].to_numpy()

figure = plt.figure(figsize=(7, 7))

plt.title('График изменения отмененных и завершенных заказов по дням')
plt.xlabel('Дни')
plt.ylabel('Количество заказов')

plt.plot(x, y, label='Отмененные заказы', marker='o', markersize=7)
plt.plot(x, y2, label='Завершенные заказы', marker='o', markersize=7)

for x1, y in zip(x, y):
    plt.text(x1, y+100, f'({x1}, {y})')

for x1, y in zip(x, y2):
    plt.text(x1 + 0.1, y - 100, f'({x1}, {y})')

plt.legend(loc='upper left')
plt.grid()
plt.show()
```

```python
y = df[df['order_status'] == 'CANCELED'].groupby(
    'order_created_day').count().reset_index()['order_status'].to_numpy()

plt.figure(figsize=(7,7))
plt.title('График завершенных и отмененных заказов по дням')
plt.xlabel('Дни')
plt.ylabel('Количество заказов')
plt.bar(x, height=y, label = 'Отмененные заказы', alpha = 0.5, width=0.5)
plt.bar(x, height=y2, label='Завершенные заказы', alpha=0.5, width=0.5)

plt.legend(loc='upper left')
plt.grid()
plt.show()
```

```python
count_order_per_store = df['store_id']

plt.figure(figsize=(7,7))
plt.title('Гистограмма')
plt.xlabel('Магазины')
plt.ylabel('Заказы')

plt.hist(count_order_per_store)
plt.grid()
plt.show()
```

```python
count_order_for_any_status = df.groupby('order_status').count().reset_index()['order_id'].to_numpy()

plt.figure(figsize=(7,7))
plt.title('Круговая диаграмма')
plt.pie(count_order_for_any_status, labels=['CANCELED', 'FINISHED'], autopct='%1.1f%%')
plt.legend()
plt.show()
```
