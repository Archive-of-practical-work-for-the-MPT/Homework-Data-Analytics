Выведем % по столбцам которые равны `null`. Столбцы `Railcard` и `Reason for Delay`, а также `Actual Arrival Time` имеют пустые значения.

```python
df.isnull().mean()*100
```

Очищать эти столбцы не имеет смысла, так как отсутсвия этих данных тоже данные. Соответсвенно их стоит заменить на более понятные текстовые данные, кроме столбца `Actual Arrival Time`, так как мы не знаем во сколько пришел поезд, то и остальные данные бессполезны, следовательно удалим `NaN`.

```python
df = df.dropna(subset=["Actual Arrival Time"]).copy()

df['Railcard'] = df['Railcard'].fillna("Not used")
df['Reason for Delay'] = df['Reason for Delay'].fillna("No delay")

df.isnull().mean() * 100
```

Проверим сколько осталось строчек после удаления `NaN` значений.

```python
df.index.size
```

Удалим все дубликаты, если они имеются. После удаления, количество строчек не поменялось, следовательно дубликатов не было.

```python
df = df.drop_duplicates()
df.shape[0]
```

Выполним срез и оставим в нем только те случаи в которых отказались от билетов. После чего посчитаем с каких станций больше всего люди отменяли билеты.

Исходя из результата топ-5 больше всего отказов в `Liverpool Lime Street`.

```python
df_group = df[['Departure Station', 'Refund Request', 'Journey Status']]

cancelled_tickets = df_group[df_group['Refund Request'].str.contains(
    'Yes', case=False)]

cancelled_tickets['Departure Station'].value_counts()[0:5]
```

Посмотрим какие уникальные значения содержит столбец `Journey Status`.

```python
df['Journey Status'].unique()
```

Посмотрим сколько уникальных значений у `Journey Status` после группировки. По результатам оказывается что только одно значение `Delayed`. А это значит что во всех случаях, когда возвращали билет поезд опаздывал.

```python
cancelled_tickets['Journey Status'].unique()
```

Выполним срез и оставим в нем только те случаи в которых поезда опоздали. После чего посчитаем на каких станциях поезда чаще всего опаздывают.

Исходя из результата топ-5 больше всего опозданий в `Liverpool Lime Street`.

```python
df_group = df[['Departure Station', 'Refund Request', 'Journey Status']]

delayed_train = df_group[df_group['Journey Status'].str.contains(
    'Delayed', case=False)]

delayed_train['Departure Station'].value_counts()[0:5]
```
